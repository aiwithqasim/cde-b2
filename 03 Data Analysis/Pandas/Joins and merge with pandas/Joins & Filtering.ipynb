## INNER JOIN

![inner join](./img/inner_join.png)

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,2],
              'Units': [40, 25, 30, 35]}
```

```python
# create a dataframe left_table_df
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table = {'CountryID': [3,4],
              'Country': ['Panama', 'Spain']}
```

```python
# create a dataframe right_table_df
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
# Inner join inner_df 

inner_df = left_table_df.merge(right_table_df,on='CountryID')
inner_df
```

```python
right_table = {'ID': [3,4],
              'Country': ['Panama', 'Spain']}
```

```python
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
inner_df = left_table_df.merge(right_table_df,left_on='CountryID',right_on='ID')
inner_df
```

```python
# with ,suffixes=['left','right'],

left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,2],
              'Units': [40, 25, 30, 35],
             'Country':['Pakistan', 'India', 'KSA', 'USA']}
```

```python
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
inner_df = left_table_df.merge(right_table_df,left_on='CountryID',right_on='ID',suffixes=['_left','_right'])
inner_df
```

## LEFT JOIN

![left join](./img/left_join.png)

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,4],
              'Units': [40, 25, 30, 35]}
```

```python
right_table = {'ID': [1,2, 3],
              'Country': ['USA', 'Canada', 'Panama']}
```

```python
right_table_df = pd.DataFrame(right_table)
left_table_df = pd.DataFrame(left_table)
```

```python
# use how
left_merged_df = left_table_df.merge(right_table_df,left_on='CountryID',right_on='ID',how='left')
left_merged_df
```

## RIGHT JOIN

![right join](./img/right_join.png)

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,4],
              'Units': [40, 25, 30, 35]}
```

```python
right_table = {'CountryID': [3, 5],
              'Country': ['Panama', 'Russia']}
```

```python
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
right_merged = left_table_df.merge(right_table_df, how='right',left_on='CountryID',right_on='CountryID')
right_merged
```

```python
# Same results using left join
right_merged = right_table_df.merge(left_table_df, how='left',left_on='CountryID',right_on='CountryID')
right_merged
```

## FULL/OUTER JOIN 

![full join](./img/full_outer_join.png)

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,2],
              'Units': [40, 25, 30, 35]}
```

```python
right_table = {'ID': [1,2,3,4],
              'Country': ['USA', 'Canada', 'Panama', 'Spain']}

```

```python
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
full_merged = left_table_df.merge(right_table_df, how='outer',left_on='CountryID',right_on='ID').drop('ID',axis=1)
full_merged
```

## Left Anti Join

![left anti](./img/left_anti_join.png)

```python
left_table = {'Date': ['01-01-2020', '02-01-2020', '03-01-2020', '04-01-2020' ],
              'CountryID': [1,1,3,2],
              'Units': [40, 25, 30, 35]}
```

```python
right_table = {'ID': [3,4],
              'Country': ['Panama', 'Spain']}
```

```python
left_table_df = pd.DataFrame(left_table)
left_table_df
```

```python
right_table_df = pd.DataFrame(right_table)
right_table_df
```

```python
left_anti_merged = left_table_df.merge(right_table_df, how='left',left_on='CountryID',right_on='ID',indicator=True).drop('ID',axis=1)
left_anti_merged
```

```python
left_anti_merged.loc[left_anti_merged['_merge'] =='left_only']
```

- suppor : **sort_index** and **sort**
